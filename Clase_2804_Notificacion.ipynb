```python
class Notificacion:
    def __init__(self, mensaje):
        self._mensaje = mensaje



class Email(Notificacion):
    def __init__(self, mensaje, destinatario_email):
        super().__init__(mensaje)

        if not isinstance(destinatario_email, str):
            raise ValueError("El destinatario_email debe ser un string válido que contenga '@'")
        if not "@" in destinatario_email:
            raise ValueError("El destinatario_email debe ser un string válido que contenga '@'")

        self.destinatario_email = destinatario_email

    def enviar(self):
        print(f"Enviando Email a {self.destinatario_email}: {self._mensaje}")


class SMS(Notificacion):
    def __init__(self, mensaje, numero):
        super().__init__(mensaje)
        self.numero = numero

    def enviar(self):
        print(f"Enviando SMS al número {self.numero}: {self._mensaje}")





```
